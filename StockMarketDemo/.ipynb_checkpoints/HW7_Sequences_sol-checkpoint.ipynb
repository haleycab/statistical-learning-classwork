# Mathematical exercises

**Exercise 1** (2 points): What is the number of scalar parameters in an HMM model with $K$ latent states and Gaussian emissions in $D$ dimensions if the emission variables are allowed to be correlated, with no particular constraint on the structure of their covariance? What if the emission variables each have independent noise, with no correlations between them?

**Solution**

Each of the $K$ states has a $D$-dimensional mean vector and either (in the first case) a covariance matrix with $D(D+1)/2$ or (in the second case) $D$ independent components associated with it. In addition, the $K\times K$ transition probability matrix has $K(K-1)$ independent components (not all components are independent since the rows must sum to 1). Thus we have (in the first case) $N_\text{params} = KD + KD(D+1)/2 + K(K-1)$ or (in the second case) $N_\text{params} = KD + KD + K(K-1)$.

## Hidden Markov models

We will use the `hmmlearn` package from here: https://github.com/hmmlearn/hmmlearn. To install the package, try running `pip install hmmlearn` in the terminal, or read the documentation.

In class, we created synthetic data from a generating model and fit the data with an HMM with two latent states and two-dimensional Gaussian emissions. We'll use the same generating model, reproduced below, for the following exercise.

```python
# Means and covariances for the two Gaussian emission distributions:
m1 = np.array([1,1])
m2 = np.array([-1,-1])
m = [m1, m2]
s1 = 0.5*np.eye(2)
s2 = 0.5*np.eye(2)
s = [s1, s2]

# trans_prob[i,j] is the probability of transitioning from state i to state j:
trans_prob = np.array([[0.8, 0.2], [0.1, 0.9]])
#trans_prob = np.array([[0.5, 0.5], [0.5, 0.5]])

n_steps = 200  # number of timesteps in our sequence

# Generate the synthetic data:
z = np.zeros(n_steps)  # latent states (0 or 1)
state = 0  # the initial state
x = np.zeros((n_steps, 2))  # observations
x[0] = np.random.multivariate_normal(m1, s1)  # initial observation
for nn in range(1, n_steps):
    state = np.random.choice(2, p=trans_prob[state])
    z[nn] = state
    x[nn,:] = np.random.multivariate_normal(m[state], s[state])
    
# Plot the results:
plt.figure(figsize=(3,3))
plt.plot(x[z==0,0], x[z==0,1], '.')
plt.plot(x[z==1,0], x[z==1,1], '.')
plt.xlabel('x1')
plt.ylabel('x2')
plt.tight_layout()

plt.figure(figsize=(8,2))
plt.plot(z, '-o')
plt.xlabel('Timesteps')
plt.ylabel('z')
plt.tight_layout()
```

As we did in class, we'll use the `hmm.GaussianHMM` class from `hmmlearn` to fit HMMs to this data.

**Exercise** (4 points): Fit HMMs to the data above as we did in class, but with different numbers $K$ of latent states. For each $K$ compute and plot the log likelihood, as well as the Bayesian information criterion (BIC), which is given by `n_params*np.log(n_steps/2) - 2*loglike`, where `n_params` is the number of parameters given in the mathematical exercise above, `n_steps` is the number of timesteps, and `loglike` is the log likelihood. Recall from the previous assignment that BIC gives models a score similar to the log likelihood except that (i) it has the opposite sign, so that smaller values are better, and (ii) it penalizes the number of parameters to prevent overfitting. In addition, for each $K$, fit a Gaussian mixture model to the data, as we did in the previous assignment, and plot the BIC for this model as well. Comment on the relationship between the two BIC curves. Is there a parameter regime in which you expect the Gaussian mixture model to be a better model than the HMM?

```python
from hmmlearn import hmm
from sklearn.mixture import GaussianMixture

## Solution ##

# Fit models to the data for different numbers of states:
ll_list = []
bic_hmm_list = []
bic_gm_list = []
kmax = 10  # maximum number of states to try
for k in range(1,kmax+1):
    # Fit the HMM model:
    model_hmm = hmm.GaussianHMM(n_components=k, covariance_type="full", n_iter=1000)
    model_hmm.fit(x[:,:])  # Fit the model to all the data
    ll_list.append(model_hmm.score(x[:,:]))
    n_params = 2*k + 2*(2+1)/2*k + k*(k-1)  # means, covs, transition matrices
    bic_hmm_list.append(n_params*np.log(n_steps/2) - 2*ll_list[-1])
    
    # Fit the Gaussian mixture model:
    model_gm = GaussianMixture(n_components=k).fit(x)
    bic_gm_list.append(model_gm.bic(x))

    
# Plot the results:
plt.figure(figsize=(4,4))
plt.subplot(211)
plt.plot(range(1,kmax+1), ll_list, '-o')
plt.ylabel('log likelihood')

plt.subplot(212)
plt.plot(range(1,kmax+1), bic_hmm_list, '-o')
plt.plot(range(1,kmax+1), bic_gm_list, '-o')
plt.ylabel('BIC')
plt.xlabel('K')
plt.legend(['HMM', 'GM'])

plt.tight_layout()
```

**Solution**

According to the BIC, when $K$ is chosen appropriately, the HMM gives the best performance, which makes sense given that it is what we used as a generative model to create the data. We see, though, that the HMM score worsens more quickly as $K$ gets large due to the fact that the larger number of parameters (in particular, the transition probabilities) are penalized. The limit in which we expect the GM model to be better than the HMM is when all transition probabilities are equal to 0.5, so that there is no longer any meaningful temporal structure in the data. It's easy to show that, in this case, the two models give nearly identical performance for small $K$, and that the GM model performs better as $K$ increases.
