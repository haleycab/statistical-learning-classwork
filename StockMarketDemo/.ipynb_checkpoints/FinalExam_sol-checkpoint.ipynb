# Coding exercises

```python
import matplotlib.pyplot as plt
import numpy as np
```

The `sklearn` library includes a dataset on breast cancer patients. For each patient, the x data is an array of measured quantities relating to the tumor, and the y data is a binary output describing whether the tumor is malignant or benign. Let's start by loading the data and looking at some of it.

```python
# Load the data:
from sklearn.datasets import load_breast_cancer
data = load_breast_cancer()

# List of the features and targets being measured:
print('Features: \n', data['feature_names'])
print('Targets: \n', data['target_names'])

# The input data (each row gives measurements of the features listed above):
X = data['data']
n_samples, n_features = np.shape(data['data'])
print('n_samples, n_features: ', (n_samples, n_features))

# The target data (whether or not the patient has cancer):
y = data['target']

# Print the first few samples of x and y data:
print('X data: \n', X[:3,:])
print('y data: \n', y[:3])

# The dataset is roughly balanced between positive and negative cases:
print('Fraction of patients with malignant tumors: ', np.sum(1-y)/len(y))
```

We will use this data to train a support vector machine to classify whether or not a patient has cancer.

To begin, notice that some features of the input data above typically contain very large values, while others typically contain very small values. It is often good practice to normalize the input data so that each feature, when measured across subjects, has a mean of zero and a standard deviation of 1. 

**Exercise**: Normalize the input data in this way, and explain why this could be an important thing to do before using the data to train a machine learning algorithm. Next, read the documentation for the `SVC` class from the `sklearn` library. Split the normalized data into training and testing data, then, using the default parameters, follow the example in the documentation to train a classifier with the training data and report its classification accuracy on the testing data.

```python
from sklearn.svm import SVC

## Solution ##

# Normalize the data:
X_means, X_stds = np.mean(X, axis=0), np.std(X, axis=0)
X_norm = (X - X_means)/X_stds

# Split the data:
X_train, X_test = X_norm[:int(0.8*n_samples)], X_norm[int(0.8*n_samples):]
y_train, y_test = y[:int(0.8*n_samples)], y[int(0.8*n_samples):]

# Train and test the classifier:
clf0 = SVC()
clf0.fit(X_train, y_train)
print('Test accuracy: ', clf0.score(X_test, y_test))
```

**Exercise**: Using the same approach as above, compare a classifier with a linear kernel against a classifier that uses radial basis functions. For the linear kernel, perfom a hyperparameter sweep of the regularization parameter `C`, plotting the test accuracy for different values of this hyperparameter. For the radial basis function kernel, plot the test accuracy on a logarithmic scale for different values of both the regularization parameter `C` and the kernel hyperparameter `gamma` (read the documentation to make sure you understand what this is). In each case, report the approximate optimal values of the hyperparameters.

For fun (it's not necessary to show this), try rerunning the solution without performing the normalization in the previous exercise, and see how it affects the results.

```python
## Solution ##

# Values of hyperparameters to try:
c_vals = 2.**np.arange(-2,12)
gamma_vals = [1e-4, 1e-3, 1e-2, 1e-1]

# Train classifiers and record the scores for each hyperparameter setting:
linear_scores = np.zeros_like(c_vals)
rbf_scores = np.zeros((len(c_vals), len(gamma_vals)))
for i, c in enumerate(c_vals):
    clf = SVC(kernel='linear', C=c, gamma=0.00001)
    clf.fit(X_train, y_train)
    linear_scores[i] = clf.score(X_test, y_test)
    for j, g in enumerate(gamma_vals):
        clf = SVC(kernel='rbf', C=c, gamma=g)
        clf.fit(X_train, y_train)
        rbf_scores[i,j] = clf.score(X_test, y_test)

# Plot the results:
plt.figure(figsize=(6,3))
plt.subplot(121)
plt.semilogx(c_vals, linear_scores, '-o')
plt.xlabel('C')
plt.ylabel('Test accuracy')

plt.subplot(122)
plt.semilogx(c_vals, rbf_scores, '-o')
plt.legend(gamma_vals)
plt.xlabel('C')
plt.tight_layout()

```

```python
print('Optimal C for linear classifier: ', c_vals[np.argmax(linear_scores)])

'''(Answers here may vary.) For the RBF classifier, the optimal combination of 
hyperparameters occurs right next to a sudden drop, so using this particular value 
might lead to a fragile classifier. A better choice might be to use values that are 
pretty good (though possibly not optimal) and also stable to small perturbations, 
e.g. (C, gamma) ~ (20, 0.01) for the curves plotted above.'''
```

The `LinearSVC` classifier is similar to the `SVC` classifier that we used above, but it can only be used with a linear kernel. Unlike the `SVC` classifier, though, it has an option to use L1 regularization. As we discussed in class, this type of regularization tends to lead to sparse solutions, so that many of the weights are equal to zero in the trained classifier.

**Exercise**: Use the `LinearSVC` classifier with arguments `penalty='l1'` and `dual=False` (read the documentation to understand what this means). Train the classifier as in the previous exercise to plot the test accuracy for different values of the regularization parameter `C`, and report the optimal value. With the optimal classifier, print the names of the features for which the weights are nonzero. These are the features of the tumor that are most predictive of whether it is malignant or benign.

```python
from sklearn.svm import LinearSVC

## Solution ##

# Train classifiers with different C values and record the test scores:
scores = []
for i, c in enumerate(c_vals):
    clf = LinearSVC(penalty='l1', dual=False, C=c)
    clf.fit(X_train, y_train)
    clf.score(X_test, y_test)
    scores.append(clf.score(X_test, y_test))

# Train a classifier with the optimal regularization parameter:
c_opt = c_vals[np.argmax(scores)]
clf = LinearSVC(penalty='l1', dual=False, C=c_opt)
clf.fit(X_train, y_train)

# Plot the test accuracy for different C:
plt.semilogx(c_vals, scores, '-o')
plt.axvline(c_opt, c='k', ls='--')
plt.xlabel('C')
plt.ylabel('Test accuracy')
    
# Report the features for which the weights are nonzero:
print('Significant features: ', data['feature_names'][clf.coef_[0]>1e-9])
```
